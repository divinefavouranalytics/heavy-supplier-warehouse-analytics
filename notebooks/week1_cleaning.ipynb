```python
import pandas as pd
```

# Week 1 Cleaning

I cleaned the date columns in the files one step at a time. I loaded each CSV from `data/raw`, changed the date fields to datetime format, and checked the data types after each change.

## Cleaning order

1. `customers.csv`: `customer_since`, `last_purchase_date`
2. `invoices.csv`: `invoice_date`, `due_date`
3. `payments.csv`: `payment_date`
4. `products.csv`: `last_purchase_date`
5. `purchase_orders_header.csv`: `order_date`, `expected_delivery_date`, `received_date`
6. `sales_orders_header.csv`: `order_date`, `delivery_date`
7. `stock_ledger.csv`: `movement_date`

## Missing date value

In `purchase_orders_header.csv`, `received_date` has 2,370 missing values. I did not fill these values because the related orders were returned, so a received date was not needed. The missing values were kept as they were.

After the date conversions, I saved the updated files in `data/cleaned`. Files that did not need changes were copied there unchanged.

```python
customers_df = pd.read_csv("../data/raw/customers.csv")


customers_df['customer_since'] = pd.to_datetime(customers_df['customer_since'])
customers_df['last_purchase_date'] = pd.to_datetime(customers_df['last_purchase_date'])

```

```python
customers_df.dtypes
```

```python
invoices_df = pd.read_csv("../data/raw/invoices.csv")


invoices_df['invoice_date'] = pd.to_datetime(invoices_df['invoice_date'])
invoices_df['due_date'] = pd.to_datetime(invoices_df['due_date'])


```

```python
invoices_df.dtypes
```

```python
payments_df = pd.read_csv("../data/raw/payments.csv")


payments_df['payment_date'] = pd.to_datetime(payments_df['payment_date'])
```

```python
payments_df.dtypes
```

```python
products_df = pd.read_csv("../data/raw/products.csv")


products_df['last_purchase_date'] = pd.to_datetime(products_df['last_purchase_date'])
```

```python
# Convert the purchase order date columns to datetime values.
purchase_orders_df = pd.read_csv("../data/raw/purchase_orders_header.csv")
purchase_orders_df['order_date'] = pd.to_datetime(purchase_orders_df['order_date'])
purchase_orders_df['expected_delivery_date'] = pd.to_datetime(purchase_orders_df['expected_delivery_date'])
purchase_orders_df['received_date'] = pd.to_datetime(purchase_orders_df['received_date'])
```

```python
purchase_orders_df.dtypes
```

```python
# Convert the sales order date columns to datetime values.
sales_orders_df = pd.read_csv("../data/raw/sales_orders_header.csv")
sales_orders_df['order_date'] = pd.to_datetime(sales_orders_df['order_date'])
sales_orders_df['delivery_date'] = pd.to_datetime(sales_orders_df['delivery_date'])
```

```python
sales_orders_df.dtypes
```

```python
# Convert the stock movement date to a datetime value.
stock_ledger_df = pd.read_csv("../data/raw/stock_ledger.csv")
stock_ledger_df['movement_date'] = pd.to_datetime(stock_ledger_df['movement_date'])
```

```python
stock_ledger_df.dtypes
```

## Save the cleaned files

I changed the date columns to date format in the seven files above. I saved those updated files in `data/cleaned`.

The other five files did not need any changes, so I copied them from `data/raw` into `data/cleaned` as well. The cleaned folder now has one complete set of 12 CSV files.

```python
import shutil

# Save the files after changing their date columns.
customers_df.to_csv("../data/cleaned/customers.csv", index=False)
invoices_df.to_csv("../data/cleaned/invoices.csv", index=False)
payments_df.to_csv("../data/cleaned/payments.csv", index=False)
products_df.to_csv("../data/cleaned/products.csv", index=False)
purchase_orders_df.to_csv("../data/cleaned/purchase_orders_header.csv", index=False)
sales_orders_df.to_csv("../data/cleaned/sales_orders_header.csv", index=False)
stock_ledger_df.to_csv("../data/cleaned/stock_ledger.csv", index=False)

# These files did not need changes, so copy them to the cleaned folder.
shutil.copy("../data/raw/branches.csv", "../data/cleaned/branches.csv")
shutil.copy("../data/raw/inventory_master.csv", "../data/cleaned/inventory_master.csv")
shutil.copy("../data/raw/purchase_orders_lines.csv", "../data/cleaned/purchase_orders_lines.csv")
shutil.copy("../data/raw/sales_orders_lines.csv", "../data/cleaned/sales_orders_lines.csv")
shutil.copy("../data/raw/suppliers.csv", "../data/cleaned/suppliers.csv")
```
