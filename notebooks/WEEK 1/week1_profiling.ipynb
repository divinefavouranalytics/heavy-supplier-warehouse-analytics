# Week 1 Data Profiling

Profile the original supplier and warehouse CSV files before cleaning.

## What I checked

I used this notebook to take a first look at the 12 CSV files in `data/raw` before cleaning them. I left the raw files as they were.

I listed the file names, loaded each file with pandas, and checked the basic details for each one:

- How many rows and columns it has
- The column names and data types
- Missing values
- Duplicate rows

This gives me a starting point for the cleaning stage. It also shows which columns may need more attention later.

```python
import pandas as pd

# List every file name here (without .csv)
file_names = [
    'branches',
    'customers',
    'inventory_master',
    'invoices',
    'payments',
    'products',
    'purchase_orders_header',
    'purchase_orders_lines',
    'sales_orders_header',
    'sales_orders_lines',
    'stock_ledger',
    'suppliers'
]
```

```python
for file in file_names:

    df = pd.read_csv(f'../data/raw/{file}.csv')

    # Print the file name clearly, so you always know which one you're looking at
    print(f"\n=========================================")
    print(f"FILE: {file}.csv")
    print(f"=========================================")
 
    print(f"Rows: {df.shape[0]}   Columns: {df.shape[1]}")
 
    print("\nColumn names and types:")
    print(df.dtypes)
 
    print("\nMissing values in each column:")
    print(df.isnull().sum())
 
    print("\nDuplicate rows (same data twice):")
    print(df.duplicated().sum())
```
